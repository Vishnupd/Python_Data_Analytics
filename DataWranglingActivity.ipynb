```python
# Inourdataset, the fuel consumption column is "mpg"
# and is represented by mpg (miles per gallon) unit.
# Assume we are developing an application in a country
# that accepts fuel consumption with the L/100km standard.
# change the name of the columnto"L/100km".

df.rename(columns={'mpg': 'L/100km'}, inplace=True)
df
```

```python
#Normalize the column “Weight” so that the values range from 0 to 1
df['weight'] = (df['weight'] - df['weight'].min()) / (df['weight'].max() - df['weight'].min())
```

```python
# In our dataset, "horsepower" is a real valued variable ranging from 48 to 288,
# it has 57 unique values. What if we only care about the price difference betweencars
# with high horsepower, medium horsepower, and little horsepower (3 types)?
# Can we rearrange them into three ‘bins' to simplify analysis?

bins = 3
df['hp_bin'] = pd.cut(df['horsepower'], bins=bins, labels=['Low','Medium','High'])
```

```python
df
```

```python
# Use a histogram to visualize the distribution of bins we created above.
import matplotlib.pyplot as plt

df['hp_bin'].value_counts().plot(kind='bar')
plt.xlabel("Horsepower Category")
plt.ylabel("Count")
plt.title("Distribution of Horsepower Bins")
plt.show()
```
